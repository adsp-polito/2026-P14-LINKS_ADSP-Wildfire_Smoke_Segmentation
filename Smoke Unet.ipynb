```python
from PIL import Image

class SmokeDataset(Dataset):

    def __init__(self, image_dir, mask_dir):

        self.image_dir = image_dir
        self.mask_dir = mask_dir

        self.images = sorted(os.listdir(image_dir))
        self.masks = sorted(os.listdir(mask_dir))

        self.transform = A.Compose([
            A.Resize(128,128)
        ])

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):

        image = cv2.imread(
            os.path.join(
                self.image_dir,
                self.images[idx]
            )
        )

        image = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB
        )

        mask = np.array(
            Image.open(
                os.path.join(
                    self.mask_dir,
                    self.masks[idx]
                )
            )
        )

        transformed = self.transform(
            image=image,
            mask=mask
        )

        image = transformed["image"]
        mask = transformed["mask"]

        image = image.astype(np.float32) / 255.0

        image = np.transpose(
            image,
            (2,0,1)
        )

        image = torch.tensor(
            image,
            dtype=torch.float32
        )

        mask = torch.tensor(
            mask,
            dtype=torch.float32
        )

        mask = mask.unsqueeze(0)

        return image, mask
```

```python
train_dataset = SmokeDataset(
    TRAIN_IMAGES,
    TRAIN_MASKS
)

valid_dataset = SmokeDataset(
    VALID_IMAGES,
    VALID_MASKS
)

test_dataset = SmokeDataset(
    TEST_IMAGES,
    TEST_MASKS
)

train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True
)

valid_loader = DataLoader(
    valid_dataset,
    batch_size=16,
    shuffle=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=16,
    shuffle=False
)
```

```python
import segmentation_models_pytorch as smp

model = smp.Unet(
    encoder_name="mit_b0",
    encoder_weights="imagenet",
    in_channels=3,
    classes=1
)
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

model = model.to(device)

print(device)
```

```python
dice_loss = smp.losses.DiceLoss(mode="binary")
focal_loss = smp.losses.FocalLoss(mode="binary")

def loss_fn(preds, targets):
    return (
        0.7 * dice_loss(preds, targets)
        + 0.3 * focal_loss(preds, targets)
    )
```

```python
loss = loss_fn(outputs, masks)

print(loss.item())
```

```python
def train_epoch(model, loader):

    model.train()

    total_loss = 0

    for images, masks in loader:

        images = images.to(device)
        masks = masks.to(device)

        optimizer.zero_grad()

        outputs = model(images)

        loss = loss_fn(outputs, masks)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    return total_loss / len(loader)
```

```python
def validate(model, loader):

    model.eval()

    total_loss = 0

    with torch.no_grad():

        for images, masks in loader:

            images = images.to(device)
            masks = masks.to(device)

            outputs = model(images)

            loss = loss_fn(outputs, masks)

            total_loss += loss.item()

    return total_loss / len(loader)
```

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=1e-4
)
```

```python
best_val = 999

for epoch in range(20):

    train_loss = train_epoch(
        model,
        train_loader
    )

    val_loss = validate(
        model,
        valid_loader
    )

    print(
        f"Epoch {epoch+1}/20 | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f}"
    )

    if val_loss < best_val:

        best_val = val_loss

        torch.save(
            model.state_dict(),
            "best_unet.pth"
        )

        print("Best model saved")
```

```python
def iou_score(preds, masks):

    preds = (preds > 0.5).float()

    intersection = (preds * masks).sum()

    union = (preds + masks).sum() - intersection

    return ((intersection + 1e-7) /
            (union + 1e-7))
```

```python
model.eval()

ious = []

with torch.no_grad():

    for images, masks in valid_loader:

        images = images.to(device)
        masks = masks.to(device)

        outputs = model(images)

        outputs = torch.sigmoid(outputs)

        score = iou_score(
            outputs,
            masks
        )

        ious.append(
            score.cpu().numpy()
        )

print("Mean IoU:", np.mean(ious))
```

```python
def f1_score(preds, masks):

    preds = (preds > 0.5).float()

    tp = (preds * masks).sum()

    precision = tp / (preds.sum() + 1e-7)

    recall = tp / (masks.sum() + 1e-7)

    f1 = (2 * precision * recall) / \
         (precision + recall + 1e-7)

    return f1
```

```python
model.eval()

f1s = []

with torch.no_grad():

    for images, masks in valid_loader:

        images = images.to(device)
        masks = masks.to(device)

        outputs = model(images)

        outputs = torch.sigmoid(outputs)

        score = f1_score(
            outputs,
            masks
        )

        f1s.append(
            score.cpu().numpy()
        )

print("Mean F1:", np.mean(f1s))
```
